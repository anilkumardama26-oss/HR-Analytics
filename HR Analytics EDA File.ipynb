BAR CHARTS-Horizontal, Grouped

```python

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
df['Attrition'].value_counts().plot(kind='bar', ax=axes[0], color=['green', 'red'])
axes[0].set_title('Employee Attrition', fontweight='bold', fontsize=12); axes[0].set_ylabel('Count')
df['Department'].value_counts().plot(kind='barh', ax=axes[1], color=['blue', 'red', 'green'])
axes[1].set_title('Department Distribution', fontweight='bold', fontsize=12); axes[1].set_xlabel('Count')
pd.crosstab(df['Department'], df['Attrition']).plot(kind='bar', ax=axes[2], color=['green', 'red'], width=0.8)
axes[2].set_title('Attrition by Department', fontweight='bold', fontsize=12); axes[2].legend(['No', 'Yes'])
plt.tight_layout(); plt.show()
```

HISTOGRAM

```python
#Age Distribution
m, md = df['Age'].mean(), df['Age'].median()
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(df['Age'], bins=15, color='purple', edgecolor='black', alpha=0.7)
ax.axvline(m, color='red', linestyle='--', linewidth=2, label=f'Mean: {m:.1f}')
ax.axvline(md, color='blue', linestyle='--', linewidth=2, label=f'Median: {md:.1f}')
ax.set(title='Age Distribution of Employees', xlabel='Age (Years)', ylabel='Frequency'); ax.grid(True, alpha=0.3); ax.legend()
plt.tight_layout(); plt.show()
```

 BOX PLOT

```python
#  Monthly Income by Job Level
fig, ax = plt.subplots(figsize=(10, 6))
df.boxplot(column='MonthlyIncome', by='JobLevel', ax=ax)
ax.set_title('Monthly Income Distribution by Job Level', fontweight='bold', fontsize=14)
ax.set_xlabel('Job Level', fontsize=12); ax.set_ylabel('Monthly Income ($)', fontsize=12)
plt.suptitle(''); plt.tight_layout(); plt.show()
```

PIE CHART

```python
#  Gender Distribution
gender = df['Gender'].value_counts()
ax = gender.plot(kind='pie', figsize=(10, 6), autopct='%1.1f%%', colors=['salmon', 'skyblue'])
ax.set_title('Gender Distribution', fontweight='bold', fontsize=14); ax.set_ylabel('')
plt.tight_layout(); plt.show()
```

HEATMAP

```python
#  Correlation Matrix
corr = df.select_dtypes(include=['number']).corr()
plt.figure(figsize=(14, 10))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True)
plt.title('Correlation Heatmap - Numerical Features', fontweight='bold', fontsize=14); plt.tight_layout(); plt.show()
```

SCATTER PLOT

```python
# Tenure vs Income
fig, ax = plt.subplots(figsize=(12, 6))
for status in ['No', 'Yes']:
    mask = df['Attrition'] == status
    ax.scatter(df[mask]['YearsAtCompany'], df[mask]['MonthlyIncome'], label=f'Attrition: {status}', alpha=0.6, s=80)
ax.set(xlabel='Years at Company', ylabel='Monthly Income ($)', title='Tenure vs Income'); ax.grid(True, alpha=0.3); ax.legend()
plt.tight_layout(); plt.show()
```
